## Data Cleaning

```python
import pandas as pd
import numpy as np
```

### Load CSV File

```python
df = pd.read_csv("netflix_titles.csv")
df.head()
```

```python
df.isnull().sum()   #check null values in every columns
```

```python
df.duplicated().sum()    #check duplicate values in dataset
```

```python
df.columns    #check column names in dataset
```

```python
df['director'] = df['director'].fillna("Unknown")     #Fill null values with some other names.
df['cast'] = df['cast'].fillna("NaN")
df['country'] = df['country'].fillna("Unknown")
df.isnull().sum()
```

```python
# Bug fix
mask = df['rating'].str.contains('min', na=False)
df.loc[mask, 'duration'] = df.loc[mask, 'rating']
df.loc[mask, 'rating'] = 'Not Rated'

# Fill remaining nulls (no dropping)
df['rating'] = df['rating'].fillna('Not Rated')
df['duration'] = df['duration'].fillna('Unknown')

# date_added
df['date_added'] = df['date_added'].fillna(df['date_added'].min())
df['date_added'] = pd.to_datetime(df['date_added'], format='%B %d, %Y')
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace('  ','_')
df.head()
```
